```python
from huggingface_hub import hf_hub_download
import pandas as pd
import duckdb
import matplotlib.pyplot as plt
```

```python
repo = "arti199919/synthetic-saas-churn-sample"

df_users = pd.read_parquet( hf_hub_download( repo_id=repo,filename="users.parquet", repo_type="dataset" ))

df_user_monthly = pd.read_parquet( hf_hub_download( repo_id=repo,  filename="user_monthly.parquet", repo_type="dataset"))
```

```python
df_users["month_signup"] = pd.to_datetime(df_users["signup_date"]).dt.strftime("%Y-%m")
```

```python
df_users_churn = duckdb.sql("""
    SELECT
        a.user_id,
        a.signup_date,
        a.month_signup,
        b.month as month_churn,
        a.churn_date,
        b.churned,
        date_diff( 'month', CAST(a.month_signup || '-01' AS DATE),CAST(b.month AS DATE) ) AS time_to_churn
    FROM df_users AS a
    LEFT JOIN df_user_monthly AS b
        ON a.user_id = b.user_id
    /*WHERE a.month_signup in ('2024-01')*/
""").df()
```

```python
df_users_churn['user_id'].nunique()
```

```python
df_users_churn[df_users_churn["churned"] == 1].sort_values("month_churn").reset_index(drop=True).head(10)
```

```python
clientes_por_churned = duckdb.sql("""
    SELECT
        month_signup,
        month_churn,
        time_to_churn,
        churned,
        COUNT(DISTINCT user_id) AS num_clientes
       /* COUNT(DISTINCT user_id) * 100.0
            / SUM(COUNT(DISTINCT user_id)) OVER (PARTITION BY month_churn) AS pct_total*/
    FROM df_users_churn
    GROUP BY 1,2,3,4
    ORDER BY 1,2,3,4
""").df()
```

```python
clientes_por_churned[clientes_por_churned['month_churn']=='2024-02-01']
```

```python
clientes_por_churned_average= duckdb.sql("""
    SELECT
        /*month_signup,*/
        month_churn,
   
        churned,
        COUNT(DISTINCT user_id) AS num_clientes,
           COUNT(DISTINCT user_id) * 100.0
            / SUM(COUNT(DISTINCT user_id)) OVER (PARTITION BY month_churn) AS churn_rate
    FROM df_users_churn
    GROUP BY 1,2
    ORDER BY 1,2
""").df()
clientes_por_churned_average.head(4)
```

```python
clientes_por_churned_t0= duckdb.sql("""
    select * from (SELECT
        /*month_signup,*/
        month_churn,
       time_to_churn,
        churned,
        COUNT(DISTINCT user_id) AS num_clientes,
           COUNT(DISTINCT user_id) * 100.0
            / SUM(COUNT(DISTINCT user_id)) OVER (PARTITION BY month_churn) AS churn_rate
    FROM df_users_churn
    where time_to_churn=0
    GROUP BY 1,2,3
    ORDER BY 1,2,3)
    where churned=1
""").df()
clientes_por_churned_t0
```

```python
clientes_churned_average = clientes_por_churned_average[clientes_por_churned_average['churned']==1]


plt.plot(clientes_churned_average['month_churn'], clientes_churned_average["churn_rate"])

plt.xlabel("Month")
plt.ylabel("Churn Rate")
plt.title("Churn Rate Over Time")
plt.ylim(0, 10)
plt.xticks(rotation=45)
plt.show()
```

```python

plt.plot(clientes_por_churned_t0['month_churn'], clientes_por_churned_t0["churn_rate"])

plt.xlabel("Month")
plt.ylabel("Churn Rate")
plt.title("Churn rate t0")
plt.ylim(0, 10)
plt.xticks(rotation=45)
plt.show()
```

```python
df=duckdb.sql("""
    SELECT
        user_id,
        MIN(month) FILTER (WHERE churned = 1) AS churn_month,
        MAX(month) AS last_month,
        MAX(month) > MIN(month) FILTER (WHERE churned = 1)
            AS appeared_after_churn
    FROM user_monthly
    GROUP BY user_id
    HAVING MIN(month) FILTER (WHERE churned = 1) IS NOT NULL
""").df()
```

```python
df = duckdb.sql("""
    SELECT
        user_id,
        MIN(month) AS first_month,
        MAX(month) AS last_month,
        date_diff('month', MIN(month), MAX(month)) AS time_to_churn
    FROM user_monthly
    GROUP BY user_id
    HAVING MAX(churned) = 1
""").df()
```
